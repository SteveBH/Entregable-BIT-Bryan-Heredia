```python
#BASE DE DATOS DESCARGADA

df = pd.read_csv("cumulative.csv")
```

```python
# Filas y columnas de la base de datos

print("Cantidad de filas y columnas:")
print(df.shape)
```

```python
print("\nColumnas:")
print(df.columns.tolist())
```

```python
# Información de la base de

print("\nInformación del dataset:")
df.info()
```

```python
# Valores Nulos

print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
# Registro de duplicados

print("\nCantidad de registros duplicados:")
print(df.duplicated().sum())
```

```python
# Limpieza de datos duplicados

df = df.drop_duplicates()

print("Cantidad de filas después de eliminar duplicados:")
print(df.shape)
```

```python
# Columna de clasificaión principal

print(df["koi_disposition"].value_counts(dropna=False))
```

```python
# eliminación de filas

df = df.dropna(subset=["koi_disposition"])

print("Filas después de eliminar valores nulos en koi_disposition:")
print(df.shape)
```

```python
columnas_analisis = [
    "koi_disposition",
    "koi_period",
    "koi_prad",
    "koi_teq",
    "koi_score"
]

print(df[columnas_analisis].isnull().sum())
```

```python
# Columnas utilizadas


datos_analisis = df[
    [
        "kepoi_name",
        "kepler_name",
        "koi_disposition",
        "koi_pdisposition",
        "koi_score",
        "koi_period",
        "koi_prad",
        "koi_teq"
    ]
].copy()

print(datos_analisis.head())
```

```python
# Casos con radio planetario registrado

datos_radio = datos_analisis.dropna(subset=["koi_prad"])

print("Registros con radio planetario disponible:")
print(datos_radio.shape)
```

```python
# Categoría por tamaño utilizando el radio del planeta

datos_analisis["planet_size"] = datos_analisis["koi_prad"].apply(
    lambda x: "Menor a 2 radios terrestres"
    if x < 2
    else "2 o más radios terrestres"
)
print(
    datos_analisis[
        ["koi_prad", "planet_size", "koi_period_years"]
    ].head()
)
```

```python
# Agrupaciones con groupBy

cantidad_clasificacion = datos_analisis.groupby(
    "koi_disposition"
).size().reset_index(name="cantidad")

print(cantidad_clasificacion)
```

```python
# Periodo orbital según clasificación

periodo_promedio = datos_analisis.groupby(
    "koi_disposition"
)["koi_period"].mean().reset_index()

print(periodo_promedio)
```

```python
# Radio promedio de los planetas clasificados

radio_promedio = datos_analisis.groupby(
    "koi_disposition"
)["koi_prad"].mean().reset_index()

print(radio_promedio)
```

```python
# Temperatura promedio de los planetas clasificados

temperatura_promedio = datos_analisis.groupby(
    "koi_disposition"
)["koi_teq"].mean().reset_index()

print(temperatura_promedio)
```

```python
# Gráficas

plt.figure(figsize=(8, 5))

sns.barplot(
    data=cantidad_clasificacion,
    x="koi_disposition",
    y="cantidad"
)

plt.title("Cantidad de objetos de Kepler por clasificación")
plt.xlabel("Clasificación")
plt.ylabel("Cantidad de objetos")

plt.tight_layout()
plt.show()
```

La gráfica muestra la cantidad de objetos de interés registrados para cada clasificación. Se observa cuál categoría concentra la mayor cantidad de registros y cuáles tienen menor representación dentro del conjunto de datos.

```python
# Gráfica de radio promedio

plt.figure(figsize=(8, 5))

plt.bar(
    radio_promedio["koi_disposition"],
    radio_promedio["koi_prad"]
)

plt.title("Radio planetario promedio según clasificación")
plt.xlabel("Clasificación")
plt.ylabel("Radio promedio (radios terrestres)")

plt.tight_layout()
plt.show()
```

La gráfica permite comparar el radio planetario promedio entre las diferentes clasificaciones. Las diferencias muestran que los objetos pertenecientes a las distintas categorías presentan diferentes valores promedio de radio.

```python
# Gráfica d periodo orbital

plt.figure(figsize=(8, 5))

plt.bar(
    periodo_promedio["koi_disposition"],
    periodo_promedio["koi_period"]
)

plt.title("Período orbital promedio según clasificación")
plt.xlabel("Clasificación")
plt.ylabel("Período orbital promedio (días)")

plt.tight_layout()
plt.show()
```

La gráfica compara el período orbital promedio entre las clasificaciones. El período orbital representa el intervalo entre tránsitos planetarios consecutivos.

```python
# Gráfica del tamano de los planetas

cantidad_tamaño = datos_analisis.groupby(
    "planet_size"
).size().reset_index(name="cantidad")

plt.figure(figsize=(8, 5))

plt.bar(
    cantidad_tamaño["planet_size"],
    cantidad_tamaño["cantidad"]
)

plt.title("Cantidad de objetos según categoría de tamaño")
plt.xlabel("Categoría de tamaño")
plt.ylabel("Cantidad de objetos")

plt.tight_layout()
plt.show()
```

La gráfica muestra cómo se distribuyen los objetos según la categoría de tamaño creada para este análisis. Permite identificar qué grupo contiene una mayor cantidad de registros.

```python
# Gráfica sobre tempratura

plt.figure(figsize=(8, 5))

plt.bar(
    temperatura_promedio["koi_disposition"],
    temperatura_promedio["koi_teq"]
)

plt.title("Temperatura de equilibrio promedio según clasificación")
plt.xlabel("Clasificación")
plt.ylabel("Temperatura promedio (K)")

plt.tight_layout()
plt.show()
```

La gráfica compara la temperatura de equilibrio promedio de los objetos según su clasificación. La temperatura de equilibrio es una estimación basada en determinados supuestos físicos sobre el planeta y la estrella anfitriona.
