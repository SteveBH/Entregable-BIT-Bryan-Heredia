1. Registro de estudiantes.

 Crea una lista llamada estudiantes. Cada elemento de la lista debe ser una tupla con la siguiente estructura:

(nombre, edad, nota) La lista debe contener mínimo 8 estudiantes.

```python
estudiantes = [
    ("Carlos", 20, 4.5),
    ("María", 22, 4.2),
    ("Andrés", 19, 3.8),
    ("Laura", 21, 4.9),
    ("Daniel", 23, 2.4),
    ("David", 23, 3.3),
    ("Diana", 24, 2.5),
    ("Juan", 27, 2.9),
]
print(estudiantes)
```

2. Mostrar la información.

 Utiliza un ciclo for para recorrer la lista e imprimir la información de cada estudiante con un formato similar al siguiente:

Ana tiene 20 años y obtuvo una nota de 4.5

```python
for i in estudiantes:
    print(f"{i[0]} tiene {i[1]} años y obtuvo una nota de {i[2]}")
```

3. Clasificación de estudiantes.

Utilizando estructuras if, elif y else, clasifica a cada estudiante según su nota.

Nota    Clasificación

Mayor o igual a 4.5    Excelente

Entre 4.0 y 4.49    Bueno

Entre 3.0 y 3.99    Aceptable

Menor a 3.0    Reprobó

Imprime el nombre del estudiante junto con su clasificación.

```python
for nombre,edad, nota in estudiantes:
  if nota >= 4.5:
    print(f"{nombre} tiene una nota excelente")
  elif nota >= 4.0 and nota <= 4.49:
      print(f"{nombre} tiene una nota buena nota")
  else:
    print(f"{nombre} tiene una nota mala")
```

4. Promedio general

Calcula el promedio de todas las notas utilizando un ciclo y una variable acumuladora.

Al finalizar imprime el promedio obtenido.

```python
suma_notas = 0

for nombre, edad, nota in estudiantes:
    suma_notas += nota

    promedio = suma_notas / len(estudiantes)

print(f"El promedio de las notas es: {promedio:.3f}")
```

5. Búsqueda de estudiante

Solicita al usuario el nombre de un estudiante mediante input().

Si el estudiante existe en la lista, imprime un mensaje indicando que fue encontrado.

Si no existe, informa que no se encontró ningún estudiante con ese nombre.

```python
nombre_buscado = input("Ingrese el nombre del estudiante a buscar: ")
encontrado = False

for nombre, edad, nota in estudiantes:
    if nombre.lower() == nombre_buscado.strip().lower():
        encontrado = True
        break

if encontrado:
    print(f"El estudiante '{nombre_buscado}' fue encontrado en la lista.")
else:
    print(f"No se encontró ningún estudiante con el nombre '{nombre_buscado}'.")
print()
```

6. Diccionario de ciudades

Crea un diccionario donde:

    • La llave sea el nombre del estudiante.

    • El valor sea la ciudad donde vive.
    
Recorre el diccionario utilizando un ciclo e imprime la información con el siguiente formato:

Ana vive en Bogotá

```python
vivienda = {
    "Carlos":"Bogotá",
    "María": "Medellín",
    "Andrés": "Pereira",
    "Laura": "Armenía",
    "Daniel": "Bogotá",
    "David": "Medellin",
    "Diana": "Bogota",
    "Juan": "Pereira",
}

for i in vivienda:
    print(f"{i} vive en {vivienda[i]}")
```

7. Cantidad de estudiantes por ciudad

A partir del diccionario anterior, construye un nuevo diccionario que indique cuántos estudiantes viven en cada ciudad.

Ejemplo:

{
    "Bogotá": 3,
    "Medellín": 2,
    "Pereira": 1
}

```python
conteo = {}

for ciudad in vivienda.values():
    if ciudad in conteo:
        conteo[ciudad] += 1
    else:
        conteo[ciudad] = 1
print(conteo)
```

8. Ciclo While.

Construye un programa que permita ingresar números utilizando un ciclo while.

El programa debe finalizar únicamente cuando el usuario ingrese el número 0.

Al finalizar muestra:

    • Cantidad de números ingresados (sin contar el 0).
    • Suma total.
    • Promedio de los números ingresados.

```python
Cantidad_numeros = 0
Suma = 0
Promedio = 0

while True:
    numero = float(input("Ingrese un número: "))
    if numero == 0:
        break
    Cantidad_numeros += 1
    Suma += numero

print(f"Cantidad de números ingresados: {Cantidad_numeros}")
print(f"Suma total: {Suma}")
print(f"Promedio de los números ingresados: {Suma / Cantidad_numeros}")
```

9. Uso de break y continue

Recorre los números del 1 al 30.

    • Si el número es múltiplo de 3, utiliza continue.
    • Si el número es igual a 25, utiliza break.
    
Imprime únicamente los números que fueron procesados por el ciclo.

```python
for i in range(1, 31):
    if i == 25:
        break
    if i % 3 == 0:
        continue
    print(i)
```
