```python
datos = datos.drop_duplicates()
```

```python
datos = datos.copy()
datos["Animal name"] = datos["Animal name"].fillna("Unknown")
```

```python
datos = datos.copy()

correcciones_animales = {
    "red squirrell": "red squirrel",
    "red squirel": "red squirrel",
    "European bison™": "European bison",
    "European bisson": "European bison",
    "European buster": "European bison",
    "ledgehod": "hedgehog",
    "wedgehod": "hedgehog",
    "lynx?": "lynx"
}

datos["Animal type"] = datos["Animal type"].replace(correcciones_animales)
```

```python
print(datos["Animal type"].value_counts(dropna=False))
```

```python
correcciones_paises = {
    "PL": "Poland",
    "HU": "Hungary",
    "Hungry": "Hungary",
    "CZ": "Czech Republic",
    "Czech": "Czech Republic",
    "DE": "Germany"
}

datos["Country"] = datos["Country"].replace(correcciones_paises)

print(datos["Country"].value_counts(dropna=False))
```

```python
print("=== ANIMAL TYPE DESPUÉS DE LA LIMPIEZA ===")
print(datos["Animal type"].value_counts(dropna=False))

print("\n=== COUNTRY DESPUÉS DE LA LIMPIEZA ===")
print(datos["Country"].value_counts(dropna=False))

print("\n=== DUPLICADOS DESPUÉS DE LA LIMPIEZA ===")
print(datos.duplicated().sum())
```

```python
peso_promedio = np.nanmean(datos["Weight kg"])

datos["Weight category"] = np.where(
    datos["Weight kg"] >= peso_promedio,
    "Above average",
    "Below average"
)

mediana_peso = np.nanmedian(datos["Weight kg"])

print("Peso promedio:", peso_promedio, "kg")
print("Mediana del peso:", mediana_peso, "kg")

print("\nEjemplo de la nueva variable:")
print(datos[["Animal type", "Weight kg", "Weight category"]].head())
```

```python
mediana_peso = np.nanmedian(datos["Weight kg"])

print("Mediana del peso:", mediana_peso, "kg")
```

4. Análisis con groupby()


Una vez que hayas terminado la limpieza, utiliza groupby() para responder preguntas sobre los datos.


Debes realizar al menos 4 análisis diferentes.


    • Cuál es el número de observaciones por tipo de animal?
    • Cuál es el promedio de una variable numérica según el tipo de animal?
    • Cuál es el promedio de una variable numérica según el país?

```python
 # 1. Cuál es el número de observaciones por tipo de animal?

observaciones_animal = datos.groupby("Animal type").size().reset_index(name="Number of observations")

print("1. Número de observaciones por tipo de animal:")
print(observaciones_animal)
```

```python
# Promedio del peso según el tipo de animal ( variable númerica según el tipo de animal)

peso_promedio_animal = datos.groupby("Animal type")["Weight kg"].mean().reset_index()

print("\n2. Peso promedio según tipo de animal:")
print(peso_promedio_animal)
```

```python
# Promedio del tamaño según el tipo de animal

tamaño_promedio_animal = datos.groupby("Animal type")["Body Length cm"].mean().reset_index()

print("\n3. Tamaño promedio según tipo de animal:")
print(tamaño_promedio_animal)
```

```python

# Promedio del peso según el país

peso_promedio_pais = datos.groupby("Country")["Weight kg"].mean().reset_index()

print("\n4. Peso promedio según país:")
print(peso_promedio_pais)
```
